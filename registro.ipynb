# Sistema de Registro de Usuarios (Versión Resumida)

Este notebook contiene una implementación esencial para validar datos de usuario (nombre, edad, correo) con manejo de excepciones y mensajes de depuración.

```python
import logging

# Configuración de logging para depuración
logging.basicConfig(level=logging.DEBUG, format='%(levelname)s: %(message)s', force=True)

def validar_datos(nombre, edad, correo):
    """
    Valida los datos de registro. 
    Requisitos: nombre no vacío, edad > 0 (entero), correo con '@'.
    """
    logging.debug(f"Iniciando validación: {nombre}, {edad}, {correo}")
    
    try:
        # 1. Validar tipos de datos (TypeError)
        if not isinstance(nombre, str) or not isinstance(correo, str):
            raise TypeError("Nombre y correo deben ser texto.")
        
        # Intentar convertir edad si es necesario
        edad = int(edad)
        
        # 2. Validar contenido (ValueError)
        if not nombre.strip():
            raise ValueError("El nombre no puede estar vacío.")
        if "@" not in correo:
            raise ValueError("El correo debe contener '@'.")
            
        # 3. Validar edad y simular ZeroDivisionError si es 0
        if edad == 0:
            logging.error("Intento de registro con edad cero.")
            _ = 1 / 0
        if edad < 0:
            raise ValueError("La edad debe ser positiva.")

        print(f"✅ Éxito: {nombre} registrado correctamente.")
        logging.info(f"Registro exitoso: {correo}")
        return True

    except TypeError as e:
        print(f"❌ Error de tipo: {e}")
    except ValueError as e:
        print(f"❌ Error de valor: {e}")
    except ZeroDivisionError:
        print("❌ Error: La edad no puede ser cero (División por cero simulada).")
    finally:
        print("--- Proceso de validación finalizado ---")
    return False
```

```python
def probar_validaciones():
    casos = [
        ("Juan Pérez", 25, "juan@mail.com"), # Válido
        ("", 30, "ana@mail.com"),        # Nombre vacío
        ("Luis", "veinte", "luis@mail.com"), # Edad no entera
        ("Marta", 0, "marta@mail.com"),    # Edad cero (ZeroDivisionError)
        ("Elena", 35, "elenamail.com"),    # Sin @
    ]
    
    for n, e, c in casos:
        print(f"\nProbando: {n}, {e}, {c}")
        validar_datos(n, e, c)

probar_validaciones()
```
