# Sistema de Registro de Usuarios con Interfaz Interactiva

Este notebook implementa un sistema de validación para el registro de usuarios con una interfaz gráfica interactiva dentro de Google Colab.

```python
import logging
import ipywidgets as widgets
from IPython.display import display, clear_output

# Configuración básica de logging para mensajes de depuración
logging.basicConfig(level=logging.DEBUG, format='%(levelname)s: %(message)s', force=True)

def validar_datos(nombre, edad, correo):
    """
    Valida los datos de registro de un usuario.
    
    Retorna: (bool, str) -> (Éxito, Mensaje informativo)
    """
    logging.debug(f"Iniciando validación para: nombre={nombre}, edad={edad}, correo={correo}")
    resultado_exitoso = False
    mensaje_retorno = ""

    try:
        # 1. Validación de tipos (TypeError)
        if not isinstance(nombre, str):
            raise TypeError("El nombre debe ser una cadena de texto.")
        if not isinstance(edad, int):
            raise TypeError("La edad debe ser un número entero.")
        if not isinstance(correo, str):
            raise TypeError("El correo debe ser una cadena de texto.")

        # 2. Validación de contenido (ValueError)
        if not nombre.strip():
            raise ValueError("El nombre no puede estar vacío.")
        
        # 3. Validación de edad (ValueError y ZeroDivisionError como ejemplo solicitado)
        if edad == 0:
            logging.debug("Edad es 0, provocando ZeroDivisionError con fines demostrativos.")
            dummy = 1 / edad 
        
        if edad < 0:
            raise ValueError("La edad debe ser un número entero mayor que cero.")

        # 4. Validación de correo (ValueError)
        if "@" not in correo:
            raise ValueError("El correo electrónico debe contener un símbolo '@'.")

        mensaje_retorno = f"Éxito: Los datos de '{nombre}' han sido validados correctamente."
        print(mensaje_retorno)
        resultado_exitoso = True

    except TypeError as e:
        logging.error(f"Error de tipo: {e}")
        mensaje_retorno = f"Error de Tipo: {e}"
    except ValueError as e:
        logging.error(f"Error de valor: {e}")
        mensaje_retorno = f"Error de Valor: {e}"
    except ZeroDivisionError as e:
        logging.error(f"Error de división por cero: {e}")
        mensaje_retorno = "Error: La edad no puede ser cero (ZeroDivisionError simulado)."
    except Exception as e:
        logging.error(f"Error inesperado: {e}")
        mensaje_retorno = f"Error inesperado: {e}"
    finally:
        print("--- Operación de validación finalizada ---")
    
    return resultado_exitoso, mensaje_retorno
```

### Interfaz Gráfica de Usuario (GUI)
Utiliza los campos a continuación para ingresar la información. Haz clic en el botón para validar.

```python
# Creación de widgets
nombre_text = widgets.Text(description="Nombre:", placeholder="Ej. Juan Perez")
edad_text = widgets.Text(description="Edad:", placeholder="Ej. 25")
correo_text = widgets.Text(description="Correo:", placeholder="Ej. juan@mail.com")
boton_validar = widgets.Button(description="Validar Registro", button_style='primary')
salida = widgets.Output()

def al_hacer_clic(b):
    with salida:
        clear_output()
        
        # Obtener valores de los widgets
        nombre = nombre_text.value
        correo = correo_text.value
        
        # Intentar procesar la edad
        try:
            if not edad_text.value.strip():
                edad = None # Forzará un TypeError en validar_datos si no es int
            else:
                edad = int(edad_text.value)
        except ValueError:
            # Si no es un entero, pasamos el string para que validar_datos dispare el TypeError
            edad = edad_text.value
        
        # Llamar a la función de validación
        exito, mensaje = validar_datos(nombre, edad, correo)
        
        if exito:
            print(f"✅ VALIDACIÓN EXITOSA: {mensaje}")
        else:
            print(f"❌ ERROR EN REGISTRO: {mensaje}")

boton_validar.on_click(al_hacer_clic)

# Mostrar la interfaz
ui = widgets.VBox([nombre_text, edad_text, correo_text, boton_validar, salida])
display(ui)
```

### Pruebas Automatizadas del Sistema
Esta sección ejecuta pruebas de estrés con diferentes tipos de datos para asegurar la robustez del sistema.

```python
def probar_validaciones():
    pruebas = [
        ("Ana Gomez", 28, "ana@test.com"),
        ("", 20, "error@test.com"),
        ("Luis", -1, "luis@test.com"),
        ("Marta", 30, "marta_at_test.com"),
        ("Elena", 0, "elena@test.com"),
        (None, "X", 123)
    ]
    
    for n, e, c in pruebas:
        print(f"\n>> Probando: {n}, {e}, {c}")
        validar_datos(n, e, c)

probar_validaciones()
```
