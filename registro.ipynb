# Sistema de Registro de Usuarios

Este notebook implementa un sistema básico de validación para el registro de usuarios, capturando nombres, edades y correos electrónicos.

```python
import logging

# Configuración básica de logging para mensajes de depuración
# Nota: En Colab, los mensajes de logging se muestran en la salida estándar.
logging.basicConfig(level=logging.DEBUG, format='%(levelname)s: %(message)s', force=True)

def validar_datos(nombre, edad, correo):
    """
    Valida los datos de registro de un usuario.
    
    Args:
        nombre (str): El nombre del usuario.
        edad (int): La edad del usuario.
        correo (str): El correo electrónico del usuario.
    """
    logging.debug(f"Iniciando validación para: nombre={nombre}, edad={edad}, correo={correo}")
    
    try:
        # 1. Validación de tipos (TypeError)
        if not isinstance(nombre, str):
            raise TypeError("El nombre debe ser una cadena de texto.")
        if not isinstance(edad, int):
            raise TypeError("La edad debe ser un número entero.")
        if not isinstance(correo, str):
            raise TypeError("El correo debe ser una cadena de texto.")

        # 2. Validación de contenido (ValueError)
        if not nombre.strip():
            raise ValueError("El nombre no puede estar vacío.")
        
        # 3. Validación de edad (ValueError y ZeroDivisionError)
        if edad == 0:
            # Forzamos un ZeroDivisionError para cumplir con el requisito de incluirlo como ejemplo
            logging.debug("Edad es 0, provocando ZeroDivisionError con fines demostrativos.")
            dummy = 1 / edad 
        
        if edad < 0:
            raise ValueError("La edad debe ser un número entero mayor que cero.")

        # 4. Validación de correo (ValueError)
        if "@" not in correo:
            raise ValueError("El correo electrónico debe contener un símbolo '@'.")

        print(f"Éxito: Los datos de {nombre} han sido validados y registrados correctamente.")
        return True

    except TypeError as e:
        logging.error(f"Error de tipo: {e}")
        print(f"Error: Tipo de dato incorrecto. {e}")
        return False
    except ValueError as e:
        logging.error(f"Error de valor: {e}")
        print(f"Error: Valor de dato no válido. {e}")
        return False
    except ZeroDivisionError as e:
        logging.error(f"Error de división por cero: {e}")
        print("Error: La edad no puede ser cero (Error de división por cero simulado).")
        return False
    except Exception as e:
        logging.error(f"Error inesperado: {e}")
        print(f"Ocurrió un error inesperado: {e}")
        return False
    finally:
        # Mensaje de cierre que siempre se muestra
        print("--- Proceso de validación finalizado ---")
```

```python
def probar_validaciones():
    """
    Función de prueba que llama a validar_datos con diferentes entradas.
    """
    pruebas = [
        ("Juan Perez", 25, "juan@example.com"),      # Válido
        ("", 30, "ana@example.com"),                 # Nombre vacío (ValueError)
        ("Carlos", -5, "carlos@example.com"),        # Edad negativa (ValueError)
        ("Marta", 22, "martaexample.com"),           # Sin @ (ValueError)
        (123, 20, "test@test.com"),                  # Nombre no string (TypeError)
        ("Luis", "veinte", "luis@test.com"),         # Edad no int (TypeError)
        ("Elena", 0, "elena@test.com"),              # Edad cero (ZeroDivisionError)
    ]

    for n, e, c in pruebas:
        print(f"\nProbando con: Nombre='{n}', Edad={e}, Correo='{c}'")
        validar_datos(n, e, c)

probar_validaciones()
```
