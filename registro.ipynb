# Sistema de Registro de Usuarios con Validación Detallada

Este notebook implementa un sistema de validación interactivo para el registro de usuarios, proporcionando retroalimentación específica para cada campo (Nombre, Edad y Correo).

```python
import logging
import ipywidgets as widgets
from IPython.display import display, clear_output

# Configuración básica de logging para mensajes de depuración
logging.basicConfig(level=logging.DEBUG, format='%(levelname)s: %(message)s', force=True)

def validar_campo(tipo, valor):
    """
    Valida un campo individual y retorna (bool, mensaje).
    Utiliza try-except para manejar las validaciones solicitadas.
    """
    try:
        if tipo == "nombre":
            if not isinstance(valor, str):
                raise TypeError("El nombre debe ser texto.")
            if not valor.strip():
                raise ValueError("El nombre no puede estar vacío.")
            return True, "✅ Correcto"
            
        elif tipo == "edad":
            if not isinstance(valor, int):
                raise TypeError("La edad debe ser un número entero.")
            if valor == 0:
                # Simulación de ZeroDivisionError según requerimiento
                dummy = 1 / valor
            if valor < 0:
                raise ValueError("La edad debe ser mayor a cero.")
            return True, "✅ Correcto"
            
        elif tipo == "correo":
            if not isinstance(valor, str):
                raise TypeError("El correo debe ser texto.")
            if "@" not in valor:
                raise ValueError("Falta el símbolo '@'.")
            return True, "✅ Correcto"
            
    except TypeError as e:
        logging.error(f"Error de tipo en {tipo}: {e}")
        return False, f"❌ Incorrecto: {e}"
    except ValueError as e:
        logging.error(f"Error de valor en {tipo}: {e}")
        return False, f"❌ Incorrecto: {e}"
    except ZeroDivisionError:
        logging.error(f"Error de división por cero en edad")
        return False, "❌ Incorrecto: La edad no puede ser cero."
    finally:
        logging.debug(f"Fin de validación individual para el campo: {tipo}")

def validar_datos(nombre, edad, correo):
    """
    Realiza la validación integral y asegura un mensaje de cierre con finally.
    """
    logging.info("Iniciando proceso de validación integral.")
    resultados = {}
    
    try:
        resultados['nombre'] = validar_campo("nombre", nombre)
        resultados['edad'] = validar_campo("edad", edad)
        resultados['correo'] = validar_campo("correo", correo)
        
        todo_valido = all(r[0] for r in resultados.values())
        
        if todo_valido:
            print(f"Registro completado con éxito para: {nombre}")
        else:
            print("Existen errores en los datos proporcionados.")
            
    except Exception as e:
        logging.error(f"Error crítico en el proceso: {e}")
        print(f"Ocurrió un error inesperado durante el registro: {e}")
    finally:
        # Requisito: Asegurar que un mensaje de cierre siempre se muestre
        print("--- Proceso de Validación Finalizado ---")
        
    return resultados

def probar_validaciones():
    """
    Función de prueba para simular errores comunes programáticamente.
    """
    print("\n--- Ejecutando Pruebas Automatizadas ---")
    pruebas = [
        ("Ana Gomez", 28, "ana@test.com"),
        ("", 20, "error@test.com"),
        ("Luis", -1, "luis@test.com"),
        ("Marta", 30, "marta_at_test.com"),
        ("Elena", 0, "elena@test.com"),
        (None, "X", 123)
    ]
    
    for n, e, c in pruebas:
        print(f"\n>> Probando Entrada: Name={n}, Age={e}, Email={c}")
        validar_datos(n, e, c)
    print("\n--- Pruebas Automatizadas Finalizadas ---")
```

### Interfaz de Registro Interactiva
Ingrese los datos solicitados. El sistema validará cada campo individualmente.

```python
# Widgets de entrada
nombre_input = widgets.Text(description="Nombre:", placeholder="Nombre completo")
edad_input = widgets.Text(description="Edad:", placeholder="Ej. 25")
correo_input = widgets.Text(description="Correo:", placeholder="usuario@ejemplo.com")

# Widgets de etiquetas para feedback
feedback_nombre = widgets.Label(value="")
feedback_edad = widgets.Label(value="")
feedback_correo = widgets.Label(value="")

boton_validar = widgets.Button(description="Validar Registro", button_style='primary')
output_general = widgets.Output()

def procesar_registro(b):
    with output_general:
        clear_output()
        
        # Preparar edad
        raw_edad = edad_input.value.strip()
        try:
            val_edad = int(raw_edad) if raw_edad else "" 
        except ValueError:
            val_edad = raw_edad
            
        # Validar
        res = validar_datos(nombre_input.value, val_edad, correo_input.value)
        
        # Actualizar feedback visual
        feedback_nombre.value = res.get('nombre', (False, "Error"))[1]
        feedback_edad.value = res.get('edad', (False, "Error"))[1]
        feedback_correo.value = res.get('correo', (False, "Error"))[1]

boton_validar.on_click(procesar_registro)

# Layout
layout = widgets.VBox([
    widgets.HBox([nombre_input, feedback_nombre]),
    widgets.HBox([edad_input, feedback_edad]),
    widgets.HBox([correo_input, feedback_correo]),
    boton_validar,
    output_general
])

display(layout)
```
